```python
url = "https://github.com/dm-fedorov/pandas_basic/blob/master/data/taxi_peru.csv.zip?raw=True"

taxi = pd.read_csv(url, compression="zip", sep=";")
taxi
```

```python
col =['start_at','end_at','arrived_at']
taxi[col] = taxi[col]\
    .apply(pd.to_datetime)
taxi.info()
```

```python
taxi = taxi[\
    taxi['start_type'].isin(['asap','reserved'])\
        ]
taxi
```

```python
taxi[['arrived_at','start_at']]
taxi['wait_time'] = taxi['arrived_at'].astype("datetime64[m]") - taxi['start_at'].astype("datetime64[m]")
taxi['wait_time']
```
